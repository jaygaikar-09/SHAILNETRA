```python
import pandas as pd

# Load dataset
df = pd.read_csv("Train.csv")

# Create a simple dataset
simple_data = pd.DataFrame()

# Keep Sample ID
simple_data["Sample_ID"] = df["Sample_ID"]

# Combine the 25 measurements of each parameter
simple_data["Elevation"] = df.filter(regex="_elevation$").mean(axis=1)
simple_data["Slope"] = df.filter(regex="_slope$").mean(axis=1)
simple_data["Aspect"] = df.filter(regex="_aspect$").mean(axis=1)
simple_data["Curvature"] = df.filter(regex="_curvature$").mean(axis=1)
simple_data["TWI"] = df.filter(regex="_twi$").mean(axis=1)
simple_data["LS_Factor"] = df.filter(regex="_ls$").mean(axis=1)
simple_data["Geology"] = df.filter(regex="_geology$").mean(axis=1)
simple_data["SDOIF"] = df.filter(regex="_sdoif$").mean(axis=1)

# Add target
simple_data["Label"] = df["Label"]

# Remove missing values
simple_data = simple_data.dropna()

# Remove duplicate rows
simple_data = simple_data.drop_duplicates()

# Save
simple_data.to_csv("simple_prediction_dataset.csv", index=False)

print("Dataset created successfully!")
print("Shape:", simple_data.shape)
print(simple_data.head())
```

```python
import pandas as pd

df = pd.read_csv("Train.csv")

print("Dataset shape:", df.shape)
print("\nAll columns:")
print(df.columns.tolist())
```

```python
print("Elevation columns:")
print([c for c in df.columns if "elevation" in c.lower()])

print("\nSlope columns:")
print([c for c in df.columns if "slope" in c.lower()])

print("\nAspect columns:")
print([c for c in df.columns if "aspect" in c.lower()])

print("\nCurvature columns:")
print([c for c in df.columns if "curvature" in c.lower()])

print("\nTWI columns:")
print([c for c in df.columns if "twi" in c.lower()])

print("\nLS columns:")
print([c for c in df.columns if "ls" in c.lower()])

print("\nGeology columns:")
print([c for c in df.columns if "geology" in c.lower()])

print("\nSDOIF columns:")
print([c for c in df.columns if "sdoif" in c.lower()])
```

```python
simple_data = pd.DataFrame()

simple_data["Sample_ID"] = df["Sample_ID"]

# Find numeric columns for each parameter
groups = {
    "Elevation": [c for c in df.columns if "elevation" in c.lower()],
    "Slope": [c for c in df.columns if "slope" in c.lower()],
    "Aspect": [c for c in df.columns if "aspect" in c.lower()],
    "Curvature": [c for c in df.columns if "curvature" in c.lower()],
    "TWI": [c for c in df.columns if "twi" in c.lower()],
    "LS_Factor": [c for c in df.columns if "ls" in c.lower()],
    "SDOIF": [c for c in df.columns if "sdoif" in c.lower()]
}

for feature, columns in groups.items():
    numeric_columns = df[columns].select_dtypes(include="number").columns
    
    if len(numeric_columns) > 0:
        simple_data[feature] = df[numeric_columns].mean(axis=1)
    else:
        print("No numeric columns found for:", feature)

# Target
simple_data["Label"] = df["Label"]

print("\nNew dataset:")
print(simple_data.head())

print("\nShape:", simple_data.shape)
```

```python
# Show exactly what columns we have
print("Available columns:")
print(simple_data.columns.tolist())

# Automatically select only columns that actually exist
wanted_features = [
    "Elevation",
    "Slope",
    "Aspect",
    "Curvature",
    "TWI",
    "LS_Factor",
    "Geology",
    "SDOIF"
]

features = [col for col in wanted_features if col in simple_data.columns]

print("\nFeatures being used:")
print(features)

print("\nMissing features:")
print([col for col in wanted_features if col not in simple_data.columns])
```

```python
X = simple_data[features]
y = simple_data["Label"]

print("X shape:", X.shape)
print("y shape:", y.shape)

print("\nFirst 5 rows:")
print(X.head())

print("\nLabels:")
print(y.value_counts())
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

model = RandomForestClassifier(
    n_estimators=200,
    random_state=42,
    class_weight="balanced"
)

model.fit(X_train, y_train)

print(" Model trained successfully!")
print("Training rows:", len(X_train))
print("Testing rows:", len(X_test))
```

```python
from sklearn.metrics import accuracy_score, classification_report

y_pred = model.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred))

print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

```python
risk_probability = model.predict_proba(X_test)[:, 1]

def risk_level(p):
    if p < 0.40:
        return "LOW"
    elif p < 0.70:
        return "MEDIUM"
    else:
        return "HIGH"

risk_levels = [risk_level(p) for p in risk_probability]
```

```python
prediction_results = X_test.copy()

prediction_results["Actual_Label"] = y_test.values
prediction_results["Predicted_Label"] = y_pred
prediction_results["Risk_Probability"] = risk_probability
prediction_results["Risk_Level"] = risk_levels

prediction_results.to_csv(
    "prediction_results.csv",
    index=False
)

print("prediction_results.csv created!")
print(prediction_results.head(10))
```
