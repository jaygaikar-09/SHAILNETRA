```python
import pandas as pd
import plotly.express as px

# Load dataset
df = pd.read_csv("simple_prediction_dataset.csv")

# Clean column names
df.columns = df.columns.str.strip()

print("Columns in dataset:")
print(df.columns.tolist())

print("\nDataset shape:")
print(df.shape)

print("\nData types:")
print(df.dtypes)
```

```python
import pandas as pd
import plotly.express as px

# Load prediction results
df = pd.read_csv("prediction_results.csv")

print("Columns:")
print(df.columns.tolist())

# Features actually present
features = [
    "Elevation",
    "Slope",
    "Aspect",
    "TWI",
    "LS_Factor",
    "SDOIF"
]

# Convert features to numeric
for col in features:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# Remove incomplete rows
df = df.dropna(subset=features)

# Use the ACTUAL column name: Actual_Label
df["Risk"] = df["Actual_Label"].map({
    0: "Stable",
    1: "Rockfall Risk"
})

print("Final dataset shape:", df.shape)
print(df[features + ["Actual_Label", "Risk"]].head())
```

```python
fig = px.scatter_3d(
    df,
    x="Elevation",
    y="Slope",
    z="SDOIF",
    color="Risk",
    hover_data=[
        "Aspect",
        "TWI",
        "LS_Factor",
        "Actual_Label"
    ],
    title="3D Rockfall Risk Visualization"
)

fig.update_layout(
    scene=dict(
        xaxis_title="Elevation",
        yaxis_title="Slope Angle",
        zaxis_title="SDOIF"
    ),
    height=700
)

fig.show()
```

```python
fig = px.scatter(
    df,
    x="Slope",
    y="Risk_Probability",
    color="Risk_Level",
    size="LS_Factor",
    hover_data=[
        "Elevation",
        "Aspect",
        "TWI",
        "SDOIF"
    ],
    title="Slope Angle vs Rockfall Risk"
)

fig.update_layout(
    xaxis_title="Slope Angle (°)",
    yaxis_title="Rockfall Risk Probability"
)

fig.show()
```

```python
risk_summary = df["Risk_Level"].value_counts().reset_index()

risk_summary.columns = ["Risk Level", "Number of Zones"]

fig = px.bar(
    risk_summary,
    x="Risk Level",
    y="Number of Zones",
    text="Number of Zones",
    title="Rockfall Risk Distribution"
)

fig.show()
```
